# Exploratory Data Analysis: Cost of Living vs Developer Salaries

This notebook explores the processed dataset to understand distributions, correlations, and key patterns before building the Streamlit dashboard.

```python
import sys

sys.path.insert(0, '..')

import pandas as pd
import plotly.express as px
```

## 1. Load Data

```python
df = pd.read_csv('../data/gold/country_affordability.csv')
print(f'Shape: {df.shape}')
df.head()
```

```python
df.dtypes
```

```python
df.describe()
```

```python
# Missing values per column
df.isnull().sum()
```

## 2. Distributions

```python
fig = px.histogram(
    df, x='cost_of_living_index', nbins=20,
    title='Distribution of Cost of Living Index',
    labels={'cost_of_living_index': 'Cost of Living Index'}
)
fig.show()
```

```python
fig = px.histogram(
    df, x='median_salary_usd', nbins=20,
    title='Distribution of Median Developer Salary (USD)',
    labels={'median_salary_usd': 'Median Salary (USD)'}
)
fig.show()
```

```python
fig = px.histogram(
    df, x='affordability_index', nbins=20,
    title='Distribution of Affordability Index',
    labels={'affordability_index': 'Affordability Index (salary / cost)'}
)
fig.show()
```

## 3. Top & Bottom Countries

```python
top10 = df.nlargest(10, 'affordability_index')
bottom10 = df.nsmallest(10, 'affordability_index')

fig = px.bar(
    top10, x='country', y='affordability_index', color='region',
    title='Top 10 Countries by Affordability Index'
)
fig.show()
```

```python
fig = px.bar(
    bottom10.sort_values('affordability_index'), x='country', y='affordability_index',
    color='region', title='Bottom 10 Countries by Affordability Index'
)
fig.show()
```

## 4. Correlation: Salary vs Cost of Living

```python
corr = df[['cost_of_living_index', 'median_salary_usd', 'affordability_index']].corr()
print('Correlation matrix:')
corr
```

```python
fig = px.scatter(
    df,
    x='cost_of_living_index',
    y='median_salary_usd',
    color='region',
    hover_name='country',
    trendline='ols',
    title='Salary vs Cost of Living (with trend line)'
)
fig.show()
```

## 5. Regional Analysis

```python
regional = (
    df.groupby('region')[['median_salary_usd', 'cost_of_living_index', 'affordability_index']]
    .mean()
    .round(1)
    .sort_values('affordability_index', ascending=False)
)
regional
```

```python
fig = px.box(
    df, x='region', y='affordability_index',
    title='Affordability Index Distribution by Region',
    points='all', hover_name='country'
)
fig.update_layout(xaxis_tickangle=-30)
fig.show()
```
